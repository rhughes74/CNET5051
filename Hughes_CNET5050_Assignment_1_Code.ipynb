#### Consider the Erdos–Renyi random graph model G(n, p) with n = 1000 nodes and edge probability p = 0.01.(a) Write down the probability distribution P(L = l) for the total number of edges L, and explain each term in your expression. Then calculate the expected number of edges, ⟨L⟩, for the specified values of n and p.

Probability Distribution:

$P(L)=  {\frac{N(N-1)}{2}\choose L}p^L (1-p)^{\frac{N(N-1)}{2}-L}$

This equation reflects the probability distribution for some probability $p$ , that there are exactly '$L$ = l' total number of edges in an Erdos–Renyi random graph model.

To explain what this equation is saying:

$p^L$ reflects the probability of drawing successes an arbitrary amount of times: $L$

$(1-p)^{\frac{N(N-1)}{2}-L}$ reflects the probability we don't draw a success for all other pairs that are possible. Or, in other words: the probability we don't make an edge over the remaining "N choose two ($\frac{N(N-1)}{2}$)" pairs leftover after $L$ pairs have been processed, and

${\frac{N(N-1)}{2}\choose L}$ represents "N choose 2, choose L" or, the picking of 2 unique nodes, an arbitrary amount of times: $L$.

#### Calculating the expected number of edges for G(1000,0.01):

$ \langle{L}\rangle = p\frac{N(N-1)}{2} $
$ \langle{L}\rangle = 0.01\frac{1000(1000-1)}{2} $

$ \langle{L}\rangle = 0.01*\frac{1000(999)}{2} $

$ \langle{L}\rangle = 0.01*499500 $

$ \langle{L}\rangle = 4995 $

#### Write down the degree distribution P(K = k) and explain each term in your expression.Calculate the expected degree, ⟨k⟩, for the specified values of n and p.

Degree Distribution:

$p_k=  {{(N-1)}\choose k}p^k (1-p)^{(N-1)-k}$

This equation reflects the probability distribution for some probability $p$ , that there are exactly '$K$ = k' total degree in an Erdos–Renyi random graph model.

To explain what this equation is saying:

$p^k$ reflects the probability of drawing successes an arbitrary amount of times: $k$

$(1-p)^{(N-1)-k}$ in a similar fashion, reflects the probability we don't draw a success for all other possibilities available.

${{(N-1)}\choose k}$ represents "N-1 choose k" or, the picking of one other node, an arbitrary amount of times: $K$.

Calculating the expected average degree for G(1000,0.01):

$ \langle{k}\rangle = p(N-1) $
$ \langle{k}\rangle = 0.01(1000-1) $

$ \langle{k}\rangle = 0.01*(999) $

$ \langle{k}\rangle = 9.99 $

#### Write a program in Python, or another programming language of your choice, that takes nand p as inputs and returns one realization of G(n, p).

```python
from itertools import combinations
import networkx as nx
import numpy as np
import scipy.stats


def create_erdos_renyi_random_graph(N, p):
    """
    Creates a Erdos-Renyi Random Graph with 'N' nodes and an edge probability, 'p'

    Parameters
    ----------
    N : int
        The number of nodes of the graph to be created.

    p : float
        The probability an edge is placed between any pair of nodes

    Returns
    -------
    G : networkx.Graph
        The resulting erdos-renyi random graph,undirected, as supported by NetworkX.

    """
    #Initialize the Graph
    from itertools import combinations
    import networkx as nx
    G = nx.Graph()
    #Initialize chance
    chance = 0

    #Add in all the nodes
    for i in range(N):
        #Add a node
        G.add_node(i)
    #Add in all the edges
    for node1_id,node2_id in combinations(G.nodes(), 2):    #For every pair of nodes within G
        chance = np.random.rand()                           #Draw a random chance 0-100%
        if chance <= p:                                     #If the chance is within the threshold 'p'
            G.add_edge(node1_id,node2_id)                   #Add an edge between that pair of nodes
    return G

N = 100
p = 0.01
print(nx.erdos_renyi_graph(100,0.01))
print(create_erdos_renyi_random_graph(N,p))
```

#### Overlay a Poisson degree distribution with the same mean degree ⟨k⟩ on the empirical degree distribution. Compare the two distributions and briefly discuss their main differences.Part 2 [50 marks]

```python
from scipy.stats import poisson
deg = list(dict(Airport.degree()).values())
#Printing the Min and Max Degree for reference purposes
print("Max Deg: ",np.max(deg))
print("Min Deg: ",np.min(deg))
avg_deg= np.mean(list(dict(Airport.degree()).values()))#Getting the average degree for the Poisson

#Starting the plotting process
fig, ax = plt.subplots()
bins = np.geomspace(np.min(deg),np.max(deg),num=200)#Making log scaled bins for the empirical data

#Note mu=lambda, k=x
k_values = range(np.min(deg),np.max(deg)+1,1)#Generating the input to the poisson PDF as all integers of k from empirical-data
poisson = poisson.pmf(k=k_values,mu=avg_deg)#Getting the P_k based on Poisson

#Plotting the Empirical and Poisson Distributions
ax.hist(deg,bins=bins,label="Airport Sample",density=True,color = 'b',histtype='step',log=True)#Plotting the Empirical
ax.loglog()#Setting to log-log scale
ax.plot(k_values,poisson,label="Poisson Distribution",color = 'r')#Plotting the Poisson

#Specifying figure details below:
handles, labels = ax.get_legend_handles_labels()#Gathering the labels to make legends per plot
ax.legend(handles=handles, labels=labels,loc="upper right",fontsize="small")#Adding legend
ax.set_title("Empirical Degree Distribution of the Airport Network vs Poisson")
ax.set_xlabel("k")
ax.set_ylim(10e-26,2)
ax.set_ylabel("Probability P_k")
plt.show()
```
